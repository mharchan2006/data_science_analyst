```python
import json, math
import pandas as pd

path = "datasets/green_spaces/parks.geojson"
with open(path, encoding="utf-8") as f:
    data = json.load(f)

def ring_area_ha(ring, lat0):
    """Approximate polygon area in hectares from lon/lat points."""
    kx = 111320 * math.cos(math.radians(lat0))
    ky = 110574
    pts = [(x * kx, y * ky) for x, y in ring]
    s = 0
    for (x1, y1), (x2, y2) in zip(pts, pts[1:] + pts[:1]):
        s += x1 * y2 - x2 * y1
    return abs(s) / 2 / 10000

def centroid(ring):
    lon = sum(c[0] for c in ring) / len(ring)
    lat = sum(c[1] for c in ring) / len(ring)
    return lat, lon

rows = []
for ft in data["features"]:
    g = ft.get("geometry") or {}
    p = ft.get("properties", {}) or {}
    tags = p.get("tags", p)  # works whether tags are nested or flat

    name = tags.get("name") or ""
    ptype = (tags.get("leisure") or tags.get("landuse")
             or tags.get("natural") or "unknown")

    t = g.get("type")
    area = None
    if t == "Point":
        lon, lat = g["coordinates"]
    elif t == "Polygon":
        ring = g["coordinates"][0]
        lat, lon = centroid(ring)
        area = ring_area_ha(ring, lat)
    elif t == "MultiPolygon":
        ring = g["coordinates"][0][0]
        lat, lon = centroid(ring)
        area = sum(ring_area_ha(poly[0], lat) for poly in g["coordinates"])
    else:
        continue  # skip lines and other geometry types

    rows.append({
        "park_name": name,
        "park_type": ptype,
        "area_ha": round(area, 4) if area is not None else None,
        "latitude": round(lat, 6),
        "longitude": round(lon, 6),
    })

parks = pd.DataFrame(rows).drop_duplicates()
parks.to_csv("datasets/green_spaces/parks.csv", index=False)

print("Total mapped green spaces:", len(parks))
print("\nBy type:")
print(parks["park_type"].value_counts())
print("\nNamed features:", (parks["park_name"] != "").sum())
print("Features with area:", parks["area_ha"].notna().sum())
parks.head(20)
```

```python
import pandas as pd
import numpy as np

parks_all = pd.read_csv("datasets/green_spaces/parks.csv")

# Approximate city centres (lat, lon) and assignment radius in km
CITIES = {
    "Cuddalore":   (11.748, 79.768, 15),
    "Puducherry":  (11.934, 79.830, 20),
   
}

def haversine(lat1, lon1, lat2, lon2):
    R = 6371
    p1, p2 = np.radians(lat1), np.radians(lat2)
    dphi = p2 - p1
    dl = np.radians(lon2 - lon1)
    a = np.sin(dphi/2)**2 + np.cos(p1)*np.cos(p2)*np.sin(dl/2)**2
    return 2 * R * np.arcsin(np.sqrt(a))

def assign_city(row):
    best, best_d = "Other (outside 3 cities)", 1e9
    for city, (clat, clon, radius) in CITIES.items():
        d = haversine(row["latitude"], row["longitude"], clat, clon)
        if d <= radius and d < best_d:
            best, best_d = city, d
    return best

parks_all["city"] = parks_all.apply(assign_city, axis=1)

def category(t):
    if t in ("park", "garden", "common", "village_green"):
        return "Park / garden"
    if t in ("wood", "forest", "nature_reserve"):
        return "Forest / woodland"
    if t in ("grass", "recreation_ground"):
        return "Grass / open ground"
    if t == "playground":
        return "Playground"
    return "Other"

parks_all["category"] = parks_all["park_type"].apply(category)

parks_all.to_csv("cleaned_data/parks_clean.csv", index=False)

print("Total features:", len(parks_all))
print("\nBy city:")
print(parks_all["city"].value_counts())
print("\nCount by city and category:")
print(pd.crosstab(parks_all["city"], parks_all["category"]))
print("\nArea (ha) by city and category:")
print(parks_all.pivot_table(index="city", columns="category",
                            values="area_ha", aggfunc="sum").round(1))
```

```python
import os
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style="whitegrid")
os.makedirs("cleaned_data", exist_ok=True)
os.makedirs("visualizations", exist_ok=True)

AIR_FILES = {
    "Cuddalore":  ("datasets/air_quality/cuddalore_raw.xlsx",  "Semmandalam (TNPCB)"),
    "Puducherry": ("datasets/air_quality/puducherry_raw.xlsx", "Jawahar Nagar (PPCC)"),
}

def load_cpcb(path, city, station):
    raw = pd.read_excel(path, header=None)
    hdr = raw.index[raw[0].astype(str).str.strip() == "From Date"][0]
    df = raw.iloc[hdr + 1:].copy()
    df.columns = [str(c).strip() for c in raw.iloc[hdr]]
    df = df.dropna(axis=1, how="all")
    df["datetime"] = pd.to_datetime(df["From Date"], format="%d-%m-%Y %H:%M", errors="coerce")
    for c in df.columns:
        if c not in ("From Date", "To Date", "datetime"):
            df[c] = pd.to_numeric(df[c], errors="coerce")
    df = df.rename(columns={"PM2.5": "pm25", "PM10": "pm10", "SO2": "so2", "NO2": "no2"})
    for k in ["pm25", "pm10", "so2", "no2"]:
        if k not in df.columns:
            df[k] = np.nan
    df["city"], df["station"] = city, station
    return df[["city", "station", "datetime", "pm25", "pm10", "so2", "no2"]]

parts = []
for city, (path, station) in AIR_FILES.items():
    if os.path.exists(path):
        parts.append(load_cpcb(path, city, station))
    else:
        print("Missing file:", path)

if not parts:
    raise FileNotFoundError("Copy the two CPCB downloads into datasets/air_quality/ as "
                            "cuddalore_raw.xlsx and puducherry_raw.xlsx, then re-run this cell.")

air_raw = pd.concat(parts, ignore_index=True).dropna(subset=["datetime"])

# Cuddalore's NO2 values (about 0.05) look implausible for ug/m3, so leave them out.
air_raw.loc[air_raw["city"] == "Cuddalore", "no2"] = np.nan

air_raw = air_raw.dropna(subset=["pm25", "pm10", "so2", "no2"], how="all")
air_raw.to_csv("cleaned_data/air_15min_clean.csv", index=False)

# Daily means per city
air_raw["date"] = air_raw["datetime"].dt.date
daily = (air_raw.groupby(["city", "station", "date"])[["pm25", "pm10", "so2", "no2"]]
         .mean().round(1).reset_index())
counts = air_raw.groupby(["city", "date"])["pm10"].count().rename("pm10_readings").reset_index()
daily = daily.merge(counts, on=["city", "date"], how="left")

# Approximate PM-based AQI using CPCB breakpoints
# (the official AQI needs 16+ hours of data per day, and 9 Nov is a partial day)
def sub_index(x, bps):
    if pd.isna(x):
        return np.nan
    for c_lo, c_hi, i_lo, i_hi in bps:
        if x <= c_hi:
            return round(i_lo + (x - c_lo) * (i_hi - i_lo) / (c_hi - c_lo))
    return 500

PM10_BP = [(0,50,0,50),(51,100,51,100),(101,250,101,200),(251,350,201,300),(351,430,301,400),(431,600,401,500)]
PM25_BP = [(0,30,0,50),(31,60,51,100),(61,90,101,200),(91,120,201,300),(121,250,301,400),(251,500,401,500)]

daily["pm10_index"] = daily["pm10"].apply(lambda v: sub_index(v, PM10_BP))
daily["pm25_index"] = daily["pm25"].apply(lambda v: sub_index(v, PM25_BP))
daily["aqi_approx"] = daily[["pm10_index", "pm25_index"]].max(axis=1)

daily.to_csv("cleaned_data/air_daily_clean.csv", index=False)
daily
```

```python
import os
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style="whitegrid")

CITIES = ["Cuddalore", "Puducherry"]

# Population by park sub_area (Census 2011). Only the sub-areas listed here are counted.
POP_BY_SUBAREA = {
    "Cuddalore":  {"Cuddalore town": 173031},
    "Puducherry": {"Puducherry": 654000},
}
WASTE_BASIS = {"Cuddalore": "municipal", "Puducherry": "UT_total"}

def read_or_none(path):
    if os.path.exists(path):
        try:
            return pd.read_csv(path)
        except pd.errors.EmptyDataError:
            return None
    return None
parks_f = read_or_none("cleaned_data/parks_clean.csv")

# Re-label every park from its coordinates, so older parks_clean.csv files work too
CENTRES = [
    ("Cuddalore",  "Cuddalore town", 11.748, 79.768, 15),
    ("Cuddalore",  "Chidambaram",    11.399, 79.693, 15),   # Chidambaram is part of Cuddalore district
    ("Puducherry", "Puducherry",     11.934, 79.830, 20),
]

def haversine(lat1, lon1, lat2, lon2):
    R = 6371
    p1, p2 = np.radians(lat1), np.radians(lat2)
    a = np.sin((p2 - p1) / 2) ** 2 + np.cos(p1) * np.cos(p2) * np.sin(np.radians(lon2 - lon1) / 2) ** 2
    return 2 * R * np.arcsin(np.sqrt(a))

def assign_area(lat, lon):
    best, best_d = ("Other (outside study area)", "Other"), 1e9
    for city, sub, clat, clon, radius in CENTRES:
        d = haversine(lat, lon, clat, clon)
        if d <= radius and d < best_d:
            best, best_d = (city, sub), d
    return best

if parks_f is not None:
    parks_f = parks_f.dropna(subset=["latitude", "longitude"]).copy()
    labels = parks_f.apply(lambda r: assign_area(r["latitude"], r["longitude"]), axis=1)
    parks_f["city"] = [l[0] for l in labels]
    parks_f["sub_area"] = [l[1] for l in labels]
    if "category" not in parks_f.columns:
        def category(t):
            if t in ("park", "garden", "common", "village_green"): return "Park / garden"
            if t in ("wood", "forest", "nature_reserve"):         return "Forest / woodland"
            if t in ("grass", "recreation_ground"):               return "Grass / open ground"
            if t == "playground":                                 return "Playground"
            return "Other"
        parks_f["category"] = parks_f["park_type"].apply(category)
    parks_f.to_csv("cleaned_data/parks_clean.csv", index=False)   # save the corrected labels for Power BI
air_f   = read_or_none("cleaned_data/air_daily_clean.csv")
waste_f = read_or_none("cleaned_data/waste_clean.csv")
water_f = None   # water is compared separately (different water types)

rows = {}
for city in CITIES:
    comp = {}

    # Air: 100 x (1 - AQI/400), using the approximate PM-based daily AQI
    if air_f is not None:
        a = air_f.loc[air_f["city"] == city, "aqi_approx"].dropna()
        if len(a):
            comp["air"] = float(np.clip(100 * (1 - a.mean() / 400), 0, 100))

    # Green: park/garden area per 1,000 residents against 9 ha per 1,000 (9 sq m per person), capped at 100
    if parks_f is not None:
        subs = POP_BY_SUBAREA[city]
        sel = parks_f[(parks_f["city"] == city) & (parks_f["sub_area"].isin(subs)) &
                      (parks_f["category"] == "Park / garden")]
        pop = sum(subs.values())
        per_1000 = sel["area_ha"].sum() / (pop / 1000)
        comp["green"] = float(min(100, per_1000 / 0.9 * 100))   # 0.9 ha per 1,000 people = 9 sq m per person
        comp["_park_ha_per_1000"] = round(per_1000, 3)

    # Waste: collection percentage, capped at 100
    if waste_f is not None:
        cw = waste_f[(waste_f["city"] == city) &
                     (waste_f["basis"] == WASTE_BASIS[city])]["collection_pct"].dropna()
        if len(cw):
            comp["waste"] = float(min(100, cw.iloc[0]))

    # Water: share of samples within limits (stays blank until water_clean.csv has real data)
    if water_f is not None and len(water_f):
        w = water_f[water_f["city"] == city].copy()
        if len(w):
            # CPCB Class C limits: pH 6.5-8.5, DO >= 4 mg/L, BOD <= 3 mg/L (missing values are not failures)
            ok = pd.DataFrame({
                "ph":  w["ph"].between(6.5, 8.5) | w["ph"].isna(),
                "do":  (w["dissolved_oxygen"] >= 4) | w["dissolved_oxygen"].isna(),
                "bod": (w["bod"] <= 3) | w["bod"].isna(),
            })
            has_any = w[["ph", "dissolved_oxygen", "bod"]].notna().any(axis=1)
            within = ok.all(axis=1) & has_any
            if has_any.any():
                comp["water"] = float(within[has_any].mean() * 100)

    rows[city] = comp

score = pd.DataFrame(rows).T
extra = score.pop("_park_ha_per_1000") if "_park_ha_per_1000" in score.columns else None
score = score.reindex(columns=["air", "green", "waste", "water"]).round(1)
score["components_used"] = score.notna().sum(axis=1)
score["green_city_score"] = score[["air", "green", "waste", "water"]].mean(axis=1).round(1)
score.to_csv("cleaned_data/green_city_score.csv")

print(score)
if extra is not None:
    print("\nPark/garden area per 1,000 residents (ha):")
    print(extra.round(3))
print("\nIllustrative index, not an official rating. Missing components are left out.")

plot = score.reset_index().rename(columns={"index": "city"}).melt(
    id_vars="city", value_vars=["air", "green", "waste", "water"]).dropna()
plt.figure(figsize=(9, 5))
ax = sns.barplot(data=plot, x="variable", y="value", hue="city")
for c in ax.containers:
    ax.bar_label(c, fmt="%.0f", padding=2)
plt.ylim(0, 110)
plt.title("Green City Score Components (0-100)")
plt.xlabel("Component")
plt.ylabel("Score")
plt.tight_layout()
plt.savefig("visualizations/green_city_score.png")
plt.show()
```

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style="whitegrid")

air   = pd.read_csv("cleaned_data/air_daily_clean.csv")
waste = pd.read_csv("cleaned_data/waste_clean.csv")
parks = pd.read_csv("cleaned_data/parks_clean.csv")

POP = {"Cuddalore": 173031, "Puducherry": 654000}
WASTE_BASIS = {"Cuddalore": "municipal", "Puducherry": "UT_total"}
palette = {"Cuddalore": "#4C72B0", "Puducherry": "#DD8452"}
order = ["Cuddalore", "Puducherry"]

# Air: mean daily PM10 (the one pollutant both stations report)
pm10 = air.groupby("city")["pm10"].mean().reindex(order)

# Waste: collection rate
coll = pd.Series({c: waste[(waste["city"] == c) & (waste["basis"] == WASTE_BASIS[c])]["collection_pct"].iloc[0]
                  for c in order})

# Green: mapped parks per 10,000 people
pk = parks[parks["category"] == "Park / garden"]
pk_count = pd.Series({c: len(pk[pk["city"] == c]) / POP[c] * 10000 for c in order})

fig, axes = plt.subplots(1, 3, figsize=(15, 5))

ax = axes[0]
ax.bar(order, pm10.values, color=[palette[c] for c in order])
ax.axhline(100, color="red", linestyle="--", label="24-h limit (100)")
for i, v in enumerate(pm10.values):
    ax.text(i, v + 2, f"{v:.0f}", ha="center")
ax.set_title("Air: mean daily PM10\n(8-9 Nov, lower is better)")
ax.set_ylabel("ug/m3")
ax.legend()

ax = axes[1]
ax.bar(order, coll.values, color=[palette[c] for c in order])
for i, v in enumerate(coll.values):
    ax.text(i, v + 1, f"{v:.1f}%", ha="center")
ax.set_ylim(0, 110)
ax.set_title("Waste: share of generated waste collected\n(higher is better)")
ax.set_ylabel("%")

ax = axes[2]
ax.bar(order, pk_count.values, color=[palette[c] for c in order], hatch="//", alpha=0.7)
for i, v in enumerate(pk_count.values):
    ax.text(i, v + 0.02, f"{v:.2f}", ha="center")
ax.set_title("Green: mapped parks per 10,000 people\n(OpenStreetMap, incomplete)")
ax.set_ylabel("parks per 10,000")

fig.suptitle("Environmental Indicators: Cuddalore vs Puducherry", fontsize=14)
plt.tight_layout()
plt.savefig("visualizations/indicator_panel.png", dpi=150)
plt.show()
print(pm10.round(1).to_dict(), coll.to_dict(), pk_count.round(2).to_dict())
```

```python
import os
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style="whitegrid")
for p in ["datasets/water_quality", "cleaned_data", "visualizations"]:
    os.makedirs(p, exist_ok=True)

# ---------- 1. Cuddalore groundwater (CGWB file) ----------
cud_raw = pd.read_csv("datasets/water_quality/cuddalore_groundwater.csv", encoding="utf-8-sig")
ec_col = [c for c in cud_raw.columns if c.startswith("Electric Conductivity")][0]   # the unit symbol is garbled in the file

cud = pd.DataFrame({
    "city": "Cuddalore",
    "water_type": "Groundwater",
    "station": cud_raw["Station"],
    "area": cud_raw["Tehsil"].str.title(),
    "latitude": cud_raw["Latitude"],
    "longitude": cud_raw["Longitude"],
    "sample_date": pd.to_datetime(cud_raw["Data Acquisition Time"], format="%d-%m-%Y %H:%M", errors="coerce"),
    "ph": cud_raw["Potential of Hydrogen (pH)"],
    "ec_us_cm": cud_raw[ec_col],
    "tds": cud_raw["Total Dissolved Solids (mg/L)"],
    "chloride": cud_raw["Chloride (mg/L)"],
    "sulphate": cud_raw["Sulphate (mg/L)"],
    "total_hardness": cud_raw["Total Hardness (mgCaCO3/L)"],
    "calcium": cud_raw["Calcium (mg/L)"],
    "magnesium": cud_raw["Magnesium (mg/L)"],
    "sodium": cud_raw["Sodium (mg/L)"],
    "fluoride": cud_raw["Fluoride (mg/L)"],
    "dissolved_oxygen": np.nan, "bod": np.nan, "cod": np.nan, "turbidity": np.nan,
    "source": "CGWB groundwater quality file (district: Cuddalore)",
})

# ---------- 2. Puducherry surface water (PPCC, September 2026) ----------
# Typed from the PDF "Surface Water Quality Data in Puducherry - September 2026". Check each value against the PDF.
pud_rows = [
    # station, lat, lon, date, ph, ec, tds, chloride, sulphate, hardness, calcium, magnesium, sodium, fluoride, do, bod, cod, turbidity
    ["Ousteri Lake",     11.950528, 79.746583, "2026-09-02", 8.14,  608,  388, 73.8,  8.86, 123.5, 22.0,  16.67,  46.1, 0.60, 8.7, 4.0, 60.0, 12.6],
    ["Kanagan Lake",     11.935104, 79.798522, "2026-09-02", 8.38, 1110,  669, 238.0, 50.02, 158.8, 33.7,  18.1,  102.4, 0.75, 9.2, 10.0, 76.0, 23.1],
    ["Chunnambar River", 11.883585, 79.787326, "2026-09-01", 7.85, 1784,  983, 292.4, 84.90, 250.9, 32.16, 41.43, 262.1, 0.70, 3.6, 2.5, 52.0, 7.6],
]
pud = pd.DataFrame(pud_rows, columns=["station", "latitude", "longitude", "sample_date", "ph", "ec_us_cm", "tds",
                                      "chloride", "sulphate", "total_hardness", "calcium", "magnesium", "sodium",
                                      "fluoride", "dissolved_oxygen", "bod", "cod", "turbidity"])
pud["sample_date"] = pd.to_datetime(pud["sample_date"])
pud["city"], pud["water_type"], pud["area"] = "Puducherry", "Surface water", "Puducherry region"
pud["source"] = "PPCC Surface Water Quality Data in Puducherry, September 2026 (NWMP)"

cols = ["city", "water_type", "station", "area", "latitude", "longitude", "sample_date", "ph", "ec_us_cm", "tds",
        "chloride", "sulphate", "total_hardness", "calcium", "magnesium", "sodium", "fluoride",
        "dissolved_oxygen", "bod", "cod", "turbidity", "source"]
water = pd.concat([cud[cols], pud[cols]], ignore_index=True)
water["sample_date"] = pd.to_datetime(water["sample_date"])

# Data-quality flag: Killai has EC 129 uS/cm next to chloride 312 mg/L, which cannot both be right
water["qc_note"] = ""
water.loc[(water["station"] == "Killai") & (water["ec_us_cm"] < 300), "qc_note"] = "EC looks inconsistent with chloride; do not use EC"
water.to_csv("datasets/water_quality/water_quality.csv", index=False)
water.to_csv("cleaned_data/water_clean.csv", index=False)

print("Records:", len(water))
print(water.groupby(["city", "water_type"]).size())
print("\nParameters available per city (non-empty values):")
print(water.groupby("city")[["ph", "ec_us_cm", "tds", "chloride", "sulphate", "total_hardness", "calcium",
                             "magnesium", "sodium", "fluoride", "dissolved_oxygen", "bod"]].count())

# ---------- 3. Compare on parameters BOTH cities have, against BIS IS 10500 acceptable limits ----------
# Cuddalore is groundwater (June 2021), Puducherry is surface water (Sept 2026): not like for like.
LIMITS = {  # parameter: (lower, upper) acceptable limit for drinking water
    "ph": (6.5, 8.5), "chloride": (None, 250), "sulphate": (None, 200),
    "total_hardness": (None, 200), "calcium": (None, 75), "magnesium": (None, 30),
}
LABELS = {"ph": "pH", "chloride": "Chloride", "sulphate": "Sulphate",
          "total_hardness": "Total hardness", "calcium": "Calcium", "magnesium": "Magnesium"}

chem = water.dropna(subset=["chloride"]).copy()           # drops the Komaratchi 2023 row (TDS only)
res = []
for p, (lo, hi) in LIMITS.items():
    ok = pd.Series(True, index=chem.index)
    if lo is not None: ok &= chem[p] >= lo
    if hi is not None: ok &= chem[p] <= hi
    tmp = chem[["city", "station"]].copy()
    tmp["parameter"], tmp["within_limit"] = LABELS[p], ok
    res.append(tmp)
res = pd.concat(res)
share = (res.groupby(["city", "parameter"])["within_limit"].mean() * 100).round(0).reset_index(name="pct_within_limit")
n = chem.groupby("city").size()
print("\nSamples used for the comparison:", n.to_dict())
print(share.pivot(index="parameter", columns="city", values="pct_within_limit"))

fig, axes = plt.subplots(1, 2, figsize=(14, 5))
ax = sns.barplot(data=share, x="parameter", y="pct_within_limit", hue="city", ax=axes[0],
                 palette={"Cuddalore": "#4C72B0", "Puducherry": "#DD8452"})
for c in ax.containers:
    ax.bar_label(c, fmt="%.0f%%", padding=2)
ax.set_ylim(0, 115)
ax.set_title("Share of samples within BIS acceptable limit")
ax.set_xlabel(""); ax.set_ylabel("% of samples"); plt.setp(ax.get_xticklabels(), rotation=20)

ax = axes[1]
sns.scatterplot(data=chem, x="total_hardness", y="chloride", hue="city", style="water_type", s=140, ax=ax,
                palette={"Cuddalore": "#4C72B0", "Puducherry": "#DD8452"})
ax.axvline(200, color="red", linestyle="--", linewidth=1)
ax.axhline(250, color="red", linestyle="--", linewidth=1)
for _, r in chem.iterrows():
    ax.annotate(r["station"], (r["total_hardness"], r["chloride"]), fontsize=8, xytext=(4, 4), textcoords="offset points")
ax.set_title("Chloride vs total hardness by station (dashed = BIS acceptable limits)")
ax.set_xlabel("Total hardness (mg/L as CaCO3)"); ax.set_ylabel("Chloride (mg/L)")
plt.tight_layout()
plt.savefig("visualizations/water_comparison.png", dpi=150)
plt.show()

# ---------- 4. Puducherry surface-water extras (DO and BOD exist only for Puducherry) ----------
pd_s = water[water["city"] == "Puducherry"].copy()
pd_s["DO_ok (>=4)"] = pd_s["dissolved_oxygen"] >= 4
pd_s["BOD_ok (<=3)"] = pd_s["bod"] <= 3
print("\nPuducherry surface water against CPCB Class C (DO >= 4 mg/L, BOD <= 3 mg/L):")
print(pd_s[["station", "dissolved_oxygen", "bod", "DO_ok (>=4)", "BOD_ok (<=3)"]].to_string(index=False))

# ---------- 5. Share of stations meeting all limits (saved for Power BI) ----------
all_ok = res.pivot_table(index=["city", "station"], columns="parameter", values="within_limit", aggfunc="all")
station_level = all_ok.all(axis=1).reset_index(name="all_within_limits")
share_city = station_level.groupby("city")["all_within_limits"].mean().mul(100).round(0)
share_city.to_csv("cleaned_data/water_share_within_bis.csv", header=["pct_stations_within_all_limits"])
print("\nStations meeting ALL six limits (%):", share_city.to_dict())
```

```python
import os
import numpy as np
import pandas as pd
import matplotlib
import matplotlib.pyplot as plt
from matplotlib.patches import FancyBboxPatch

# ---------------- settings ----------------
BLUE, ORANGE, GREEN, GREY = "#3B6EA8", "#C9701B", "#2E7D5B", "#9AA5A0"
INK, BG, DARK = "#16302A", "#F6F5EF", "#12332B"
COL = {"Cuddalore": BLUE, "Puducherry": ORANGE}
CITIES = ["Cuddalore", "Puducherry"]
PUDU_WASTE_BASIS = "PUAA_plan"      # use "UT_total" only if you can cite the 499 / 470 source
MUNICIPAL_PARKS = 62                # parks reported by Cuddalore Municipality
os.makedirs("visualizations", exist_ok=True)

def read(path):
    return pd.read_csv(path) if os.path.exists(path) else None

air15 = read("cleaned_data/air_15min_clean.csv")
daily = read("cleaned_data/air_daily_clean.csv")
parks = read("cleaned_data/parks_clean.csv")
waste = read("cleaned_data/waste_clean.csv")
water = read("cleaned_data/water_clean.csv")

# Cuddalore includes Chidambaram; keep only the two study cities
if parks is not None:
    parks["city"] = parks["city"].replace({"Chidambaram": "Cuddalore"})
    parks = parks[parks["city"].isin(CITIES)].copy()
    if "category" not in parks.columns:
        parks["category"] = np.where(parks["park_type"].isin(["park", "garden", "common", "village_green"]),
                                     "Park / garden", "Other")

# ---------------- numbers for the KPI cards ----------------
def safe(fn, default="n/a"):
    try:
        return fn()
    except Exception:
        return default

pm10 = safe(lambda: daily.groupby("city")["pm10"].mean().reindex(CITIES))
park_n = safe(lambda: parks[parks["category"] == "Park / garden"].groupby("city").size().reindex(CITIES).fillna(0))

def waste_rate(city):
    basis = "municipal" if city == "Cuddalore" else PUDU_WASTE_BASIS
    return waste[(waste["city"] == city) & (waste["basis"] == basis)]["collection_pct"].dropna().iloc[0]
waste_k = {c: safe(lambda c=c: waste_rate(c)) for c in CITIES}

LIMITS = {"pH": ("ph", 6.5, 8.5), "Chloride": ("chloride", None, 250), "Sulphate": ("sulphate", None, 200),
          "Total hardness": ("total_hardness", None, 200), "Calcium": ("calcium", None, 75),
          "Magnesium": ("magnesium", None, 30)}
share, all_ok = None, None
if water is not None:
    chem = water.dropna(subset=["chloride"]).copy()
    rows, flags = [], pd.DataFrame(index=chem.index)
    for label, (col, lo, hi) in LIMITS.items():
        ok = pd.Series(True, index=chem.index)
        if lo is not None: ok &= chem[col] >= lo
        if hi is not None: ok &= chem[col] <= hi
        flags[label] = ok
        for city in CITIES:
            sel = chem["city"] == city
            rows.append({"city": city, "parameter": label, "pct": ok[sel].mean() * 100})
    share = pd.DataFrame(rows)
    flags["city"], flags["station"] = chem["city"], chem["station"]
    all_ok = flags.groupby(["city", "station"])[list(LIMITS)].all().all(axis=1).groupby("city").mean() * 100

def fmt(v, suffix="", nd=0):
    return "n/a" if isinstance(v, str) or pd.isna(v) else f"{v:.{nd}f}{suffix}"

# ---------------- layout ----------------
matplotlib.rcParams.update({"font.size": 11, "axes.edgecolor": "#CBD3CE", "axes.titleweight": "bold",
                            "axes.titlesize": 13, "axes.facecolor": "white"})
fig = plt.figure(figsize=(20, 11.25), facecolor=BG)

# header band
fig.patches.append(FancyBboxPatch((0, 0.915), 1, 0.085, transform=fig.transFigure, boxstyle="square,pad=0",
                                  facecolor=DARK, edgecolor="none"))
fig.text(0.025, 0.962, "Green City Analysis: Cuddalore vs Puducherry", color="white", fontsize=27, fontweight="bold", va="center")
fig.text(0.025, 0.928, "Environmental comparison dashboard (Cuddalore includes Chidambaram)", color="#C9D6CF", fontsize=14, va="center")
for i, c in enumerate(CITIES):
    fig.text(0.80 + i * 0.09, 0.955, "\u25A0 " + c, color=COL[c], fontsize=13, fontweight="bold", va="center",
             bbox=dict(facecolor="white", edgecolor="none", boxstyle="round,pad=0.3"))

# KPI cards
cards = [
    ("Mean PM10 (ug/m3, limit 100)", fmt(pm10["Cuddalore"] if not isinstance(pm10, str) else pm10), fmt(pm10["Puducherry"] if not isinstance(pm10, str) else pm10)),
    ("Mapped parks and gardens", fmt(park_n["Cuddalore"] if not isinstance(park_n, str) else park_n), fmt(park_n["Puducherry"] if not isinstance(park_n, str) else park_n)),
    ("Waste collection rate", fmt(waste_k["Cuddalore"], "%", 1), fmt(waste_k["Puducherry"], "%", 1)),
    ("Water stations within all 6 BIS limits", fmt(all_ok["Cuddalore"], "%") if all_ok is not None else "n/a",
     fmt(all_ok["Puducherry"], "%") if all_ok is not None else "n/a"),
]
cw, gap, x0, y0, ch = 0.2255, 0.0165, 0.025, 0.775, 0.115
for i, (title, a, b) in enumerate(cards):
    x = x0 + i * (cw + gap)
    fig.patches.append(FancyBboxPatch((x, y0), cw, ch, transform=fig.transFigure, boxstyle="round,pad=0.002,rounding_size=0.01",
                                      facecolor="white", edgecolor="#DDE3DE"))
    fig.text(x + 0.012, y0 + ch - 0.022, title, fontsize=12.5, fontweight="bold", color=INK, va="center")
    fig.text(x + cw * 0.27, y0 + 0.04, a, fontsize=30, fontweight="bold", color=BLUE, ha="center", va="center")
    fig.text(x + cw * 0.73, y0 + 0.04, b, fontsize=30, fontweight="bold", color=ORANGE, ha="center", va="center")
    fig.text(x + cw * 0.27, y0 + 0.012, "Cuddalore", fontsize=10.5, color="#5E6B64", ha="center")
    fig.text(x + cw * 0.73, y0 + 0.012, "Puducherry", fontsize=10.5, color="#5E6B64", ha="center")

gs = fig.add_gridspec(2, 3, left=0.04, right=0.985, bottom=0.14, top=0.735, wspace=0.22, hspace=0.5)

def panel(ax, title, fn):
    ax.set_title(title, loc="left", color=INK)
    try:
        fn(ax)
    except Exception as e:
        ax.text(0.5, 0.5, "Data not available", ha="center", va="center", color=GREY, transform=ax.transAxes)
        ax.set_xticks([]); ax.set_yticks([])
    for s in ("top", "right"):
        ax.spines[s].set_visible(False)

# 1 daily PM10
def p1(ax):
    d = daily.copy(); d["date"] = d["date"].astype(str)
    dates = sorted(d["date"].unique()); w = 0.38
    for i, c in enumerate(CITIES):
        vals = [d[(d["city"] == c) & (d["date"] == x)]["pm10"].mean() for x in dates]
        bars = ax.bar(np.arange(len(dates)) + (i - 0.5) * w, vals, w, color=COL[c], label=c)
        for r, v in zip(bars, vals):
            if pd.notna(v): ax.text(r.get_x() + r.get_width() / 2, v + 2, f"{v:.0f}", ha="center", fontsize=10)
    ax.axhline(100, color="red", ls="--", lw=1, label="24-h limit (100)")
    ax.set_xticks(range(len(dates))); ax.set_xticklabels(dates)
    ax.set_ylabel("ug/m3"); ax.legend(frameon=False, fontsize=9, loc="upper center", ncol=3, bbox_to_anchor=(0.5, -0.12))
panel(fig.add_subplot(gs[0, 0]), "Air: daily mean PM10", p1)

# 2 hourly PM10
def p2(ax):
    a = air15.copy(); a["datetime"] = pd.to_datetime(a["datetime"]); a["h"] = a["datetime"].dt.floor("h")
    for c in CITIES:
        h = a[a["city"] == c].groupby("h")["pm10"].mean()
        ax.plot(h.index, h.values, color=COL[c], lw=2, label=c)
    ax.axhline(100, color="red", ls="--", lw=1)
    ax.set_ylabel("ug/m3"); ax.legend(frameon=False, fontsize=9, loc="upper left")
    ax.tick_params(axis="x", labelrotation=30, labelsize=9)
panel(fig.add_subplot(gs[0, 1]), "Air: hourly PM10 (evening spike on 8 Nov)", p2)

# 3 waste collection rates
def p3(ax):
    w = waste.dropna(subset=["collection_pct"]).copy()
    names = {"municipal": "Municipality", "PUAA_plan": "Urban agglom.\n(plan, c.2016-17)", "UT_total": "UT, 5 bodies\n(source?)"}
    w["label"] = w["city"] + "\n" + w["basis"].map(names).fillna(w["basis"])
    bars = ax.bar(w["label"], w["collection_pct"], color=[COL.get(c, GREY) for c in w["city"]])
    for r, v in zip(bars, w["collection_pct"]):
        ax.text(r.get_x() + r.get_width() / 2, v + 1.5, f"{v:.1f}%", ha="center", fontsize=10)
    ax.set_ylim(0, 112); ax.set_ylabel("% collected"); ax.tick_params(axis="x", labelsize=9)
panel(fig.add_subplot(gs[0, 2]), "Waste: collection rate (compare rates, not tonnes)", p3)

# 4 parks
def p4(ax):
    labels = ["Cuddalore\n(OpenStreetMap)", "Puducherry\n(OpenStreetMap)", "Cuddalore\n(Municipality reports)"]
    vals = [park_n["Cuddalore"], park_n["Puducherry"], MUNICIPAL_PARKS]
    bars = ax.bar(labels, vals, color=[BLUE, ORANGE, GREEN])
    for r, v in zip(bars, vals): ax.text(r.get_x() + r.get_width() / 2, v + 1.5, f"{v:.0f}", ha="center", fontsize=10)
    ax.set_ylabel("parks and gardens"); ax.tick_params(axis="x", labelsize=9)
panel(fig.add_subplot(gs[1, 0]), "Green: mapped parks (OSM is incomplete)", p4)

# 5 water share within limits
def p5(ax):
    pv = share.pivot(index="parameter", columns="city", values="pct").reindex(list(LIMITS))
    x = np.arange(len(pv)); w = 0.38
    for i, c in enumerate(CITIES):
        bars = ax.bar(x + (i - 0.5) * w, pv[c], w, color=COL[c], label=c)
        for r, v in zip(bars, pv[c]): ax.text(r.get_x() + r.get_width() / 2, v + 2, f"{v:.0f}", ha="center", fontsize=8.5)
    ax.set_xticks(x); ax.set_xticklabels(pv.index, rotation=20, ha="right", fontsize=9)
    ax.set_ylim(0, 118); ax.set_ylabel("% of samples"); ax.legend(frameon=False, fontsize=9, ncol=2, loc="upper center")
panel(fig.add_subplot(gs[1, 1]), "Water: samples within BIS acceptable limit", p5)

# 6 water scatter
def p6(ax):
    for c in CITIES:
        s = chem[chem["city"] == c]
        ax.scatter(s["total_hardness"], s["chloride"], s=90, color=COL[c], label=c,
                   marker="o" if c == "Cuddalore" else "X")
        for _, r in s.iterrows():
            ax.annotate(r["station"], (r["total_hardness"], r["chloride"]), fontsize=7.5, xytext=(4, 4), textcoords="offset points")
    ax.margins(x=0.12, y=0.08)
    ax.axvline(200, color="red", ls="--", lw=1); ax.axhline(250, color="red", ls="--", lw=1)
    ax.set_xlabel("Total hardness (mg/L)"); ax.set_ylabel("Chloride (mg/L)"); ax.legend(frameon=False, fontsize=9)
panel(fig.add_subplot(gs[1, 2]), "Water: hardness vs chloride by station", p6)

# footer
fig.text(0.025, 0.05, "Sources: CPCB/TNPCB/PPCC continuous monitoring (8-9 Nov 2026), OpenStreetMap, Cuddalore Municipality, "
         "Puducherry SWM action plan, CGWB groundwater (Cuddalore, 2021), PPCC surface water (Puducherry, Sept 2026).",
         fontsize=10, color="#5E6B64")
fig.text(0.025, 0.022, "Cautions: air is a two-day snapshot including a festival evening; OSM parks are incomplete; waste sources differ in date and area; "
         "water compares groundwater with surface water (hardness is largely natural).", fontsize=10, color="#5E6B64")

fig.savefig("visualizations/dashboard.png", dpi=150, facecolor=BG)
plt.show()
print("Saved: visualizations/dashboard.png")
```
